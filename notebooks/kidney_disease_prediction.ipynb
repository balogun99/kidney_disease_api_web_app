# Importing Libraries

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score, classification_report
import seaborn as sns
import matplotlib.pyplot as plt
```

# Data Collection & Analysis

## Load Kidney Disease Dataset

```python
kidney_data = pd.read_csv('/Users/macbookpro/Desktop/End-to-End Projects/kidney_disease/data/kidney_disease_dataset.csv')
kidney_data
```

```python
# check the distribution of the Dialysis_Needed column
kidney_data['Dialysis_Needed'].value_counts()
```

## 0 ==> Dialysis ABSENT
## 1 ==> Dialysis PRESENT

```python
# group the last column by mean
kidney_data.groupby('Dialysis_Needed').mean()
```

```python
# separate FEATURES from the TARGET
X = kidney_data.drop(columns='Dialysis_Needed', axis=1)
Y = kidney_data['Dialysis_Needed']
```

```python
print(X)
```

```python
print(Y)
```

```python
# print the columns length and details
print(X.columns)
print(len(X.columns))
```

# Data Splitting

### Train Test Split

```python
# train test split
X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.4, stratify=Y, random_state=42
)
```

```python
# no of X train and test shape
print(X.shape, X_train.shape, X_test.shape)
```

# Model Training

```python
# train the model using SVC
classifier = svm.SVC(kernel='linear')
```

```python
# fit the model
classifier.fit(X_train, y_train)
```

# Model Evaluation

## Accuracy Score on Training Data

```python
# train data
X_train_pred = classifier.predict(X_train)
train_acc = accuracy_score(X_train_pred, y_train)
```

```python
print("Result on Training Data: ",train_acc)
```

## Accuracy Score on Testing Data

```python
X_test_pred = classifier.predict(X_test)
test_acc = accuracy_score(X_test_pred, y_test)
```

```python
print("Result on Testing Data: ", test_acc)
```

# Predictive System

```python
# load the dataset data
input_data = (71,0.3,40.9,0,1,46.8,1622,1) # Dialysis Absent
input_data = (42,3.14,50.3,1,1,120,2656,1) # Dialysis Present


# change the input data into a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the array
input_reshaped = input_data_as_numpy_array.reshape(1,-1)

# make predictions
predictions = classifier.predict(input_reshaped)
print(predictions)

# system check
if (predictions[0] == 1):
    print('DIALYSIS ABSENT')
else:
    print("DIALYSIS PRESENT")
```
